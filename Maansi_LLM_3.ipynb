```python
import gradio as gr
import torch
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    GPT2Tokenizer,
    GPT2LMHeadModel,
    pipeline,
    TextGenerationPipeline
)

# FinBERT (Sentiment Analysis for Finance)
finbert_tokenizer = AutoTokenizer.from_pretrained("ProsusAI/finbert")
finbert_model = AutoModelForSequenceClassification.from_pretrained("ProsusAI/finbert")
finbert_pipeline = pipeline("sentiment-analysis", model=finbert_model, tokenizer=finbert_tokenizer)

# GPT-2 (General-Purpose Foundation Model)
gpt2_tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
gpt2_model = GPT2LMHeadModel.from_pretrained("gpt2")
gpt2_pipeline = TextGenerationPipeline(model=gpt2_model, tokenizer=gpt2_tokenizer, device=0 if torch.cuda.is_available() else -1)

def simulate_financial_generation(input_text):
    finance_prompt = "Financial Report: " + input_text
    result = gpt2_pipeline(finance_prompt, max_length=60, num_return_sequences=1, do_sample=True, temperature=0.7)
    return result[0]['generated_text']


def compare_models(user_input):
    # FinBERT sentiment
    sentiment = finbert_pipeline(user_input)[0]
    sentiment_result = f"{sentiment['label']} (Confidence: {sentiment['score']:.2f})"

    # GPT-2 generation (General)
    gpt2_output = gpt2_pipeline(user_input, max_length=60, num_return_sequences=1, do_sample=True, temperature=0.7)[0]['generated_text']

    # FinGPT (Domain-specific via prompt)
    fingpt_output = simulate_financial_generation(user_input)

    return sentiment_result, gpt2_output, fingpt_output

interface = gr.Interface(
    fn=compare_models,
    inputs=gr.Textbox(lines=4, label="Enter Financial Text or Prompt", placeholder="e.g. Apple stock rises after earnings report..."),
    outputs=[
        gr.Textbox(label="📊 FinBERT Sentiment"),
        gr.Textbox(label="🤖 GPT-2 Output (General Foundation)"),
        gr.Textbox(label="💹 Simulated FinGPT Output (Finance Style)")
    ],
    title="🧠 Comparative Analysis: Foundation vs Finance Domain-Specific Models",
    description=(
        "This tool compares FinBERT (for financial sentiment) with GPT-2 (foundation model) and "
        "a simulated FinGPT using GPT-2 with financial-style prompt. Enter any finance-related news, headline, or question."
    ),
    theme="default",
    allow_flagging="never"
)

interface.launch(share=True)
```
